<h1>Extraction temporaire des données zips</h1>

```python
#Importation des librairies nécéssaire
import zipfile
import tempfile
import os
```

```python
#Les fichiers à dézipé
first_path_zip= r"Datasets/EmotionTexteDataset.zip"
two_path_zip= r"Datasets/SentimentDataset.zip"

base_dir = os.path.dirname(os.path.abspath("Datasets"))
```

```python
#Création d'un dossier temporaire pour stoker les données dézipé
with tempfile.TemporaryDirectory(dir = base_dir) as temp_dir:
    print(f"Dossier temporaire : {temp_dir}")

    #Les dossier pour chaque archive pour nos donnée
    extractData1 = os.path.join(temp_dir, "file1")
    extractData2 = os.path.join(temp_dir, "file2")

    os.makedirs(extractData1, exist_ok = True)
    os.makedirs(extractData2, exist_ok = True)

    #On dézip l'EmotionTextDataset
    with zipfile.ZipFile(first_path_zip, "r") as z:
        z.extractall(extractData1)
    #On dézip SentimentDataset
    with zipfile.ZipFile(two_path_zip, "r") as z:
        z.extractall(extractData2)
    print("Les fichiers sont dézippés de manière temporaire.")

    input("Appuie sur la touche 'Entrée' pour terminée le procéssus.")
```

```python
print("Les fichiers temporaires ont été supprimés.")
```
